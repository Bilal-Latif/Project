```python
# Function to take input for the chromosomes
def get_population_input(num_chromosomes, num_genes):
    population = []
    print(f"Enter {num_chromosomes} chromosomes, each with {num_genes} genes (0 or 1):")
    for i in range(num_chromosomes):
        chromosome = []
        while len(chromosome) < num_genes:
            gene = input(f"Enter gene {len(chromosome) + 1} for chromosome {i + 1} (0 or 1): ")
            if gene in ['0', '1']:
                chromosome.append(int(gene))
            else:
                print("Invalid input. Please enter 0 or 1.")
        population.append(chromosome)
    return population

# Function to calculate fitness (number of 1s in each chromosome)
def calfitness(S):
    fit = []
    total = 0
    print("Fitness:")
    for i in S:
        fit.append(i.count(1))  # Count the number of 1s in each individual
        total += i.count(1)
    print(fit)
    print(total)
    return fit

# Main Code
if __name__ == "__main__":
    # Number of chromosomes and genes
    num_chromosomes = 4  # Fixed number of chromosomes
    num_genes = 3        # Fixed number of genes per chromosome
    
    # Get the initial population (chromosomes)
    S = get_population_input(num_chromosomes, num_genes)
    print("Initial Population (S):", S)

    # Calculate fitness of the initial population
    fit = calfitness(S)

    # Sorting the population in descending order based on fitness
    print("Arranging in Descending order")
    desc = S[:]
    for i in range(len(desc)):
        for j in range(i + 1, len(desc)):
            if desc[i].count(1) < desc[j].count(1):
                desc[i], desc[j] = desc[j], desc[i]

    print("s =", desc)

    # Crossover (2-point crossover)
    print("Crossover after 2 points")
    # Crossover between S1 (desc[0]) and S4 (desc[3])
    for i in range(2):
        desc[0][i], desc[3][i] = desc[3][i], desc[0][i]  # Crossover between first and fourth individual
    # Crossover between S5 (desc[4]) and S6 (desc[5]) -- not possible with only 4 chromosomes, so we skip

    # Swap the positions of the crossed individuals
    desc[0], desc[3] = desc[3], desc[0]

    # Printing the result after crossover
    print("s1 and s4")
    print("s1 =", desc[0], "s4 =", desc[3])
    print("s =", desc)

    # New Crossover between S2 (desc[1]) and S3 (desc[2]) after the 4th point (if possible)
    print("Crossover between S2 and S3 after the 4th point")
    for i in range(2, num_genes):  # Perform crossover after 3rd point (index 2)
        desc[1][i], desc[2][i] = desc[2][i], desc[1][i]

    # Printing the result after this new crossover
    print("s2 and s3")
    print("s2 =", desc[1], "s3 =", desc[2])
    print("s =", desc)

    # Mutation (flip bits in the first two positions of each individual)
    print("Mutation")
    for i in range(len(desc)):
        for j in range(2):  # Mutating the first two bits
            if desc[i][j] == 0:
                desc[i][j] = 1
            else:
                desc[i][j] = 0

    # Printing the result after mutation
    print("s =", desc)

    # Recalculate fitness of the new population after mutation
    fit2 = calfitness(desc)

    # Comparing fitness before and after applying Genetic Algorithm
    if sum(fit2) > sum(fit):
        print("Fitness is greater after applying Genetic Algorithm")
    else:
        print("Fitness is greater before applying Genetic Algorithm")
```
